### Check for leading/trailing spaces
Check categorical columns for unwanted leading or trailing spaces.

Although the datasets do not contain missing values or duplicate rows, inconsistent spaces in categorical fields can create incorrect categories during grouping and analysis.

Text fields are therefore cleaned using string trimming.

```python
employee_clean = employee.copy()
performance_clean = performance.copy()
education_clean = education.copy()
rating_clean = rating.copy()
satisfaction_clean = satisfaction.copy()
```

```python
# Clean text columns in Employee table
categorical_employee = employee_clean.select_dtypes(include='str').columns

for col in categorical_employee:
    employee_clean[col] = employee_clean[col].str.strip()


# Clean text columns in Performance table
categorical_performance = performance_clean.select_dtypes(include='str').columns

for col in categorical_performance:
    performance_clean[col] = performance_clean[col].str.strip()
```

```python
education_clean["EducationLevel"] = education_clean["EducationLevel"].astype(str).str.strip()
rating_clean["RatingLevel"] = rating_clean["RatingLevel"].astype(str).str.strip()
satisfaction_clean["SatisfactionLevel"] = satisfaction_clean["SatisfactionLevel"].astype(str).str.strip()
```

```python
print("Employee text columns:")
print(employee_clean.select_dtypes(include='str').columns.tolist())

print("\nPerformance text columns:")
print(performance_clean.select_dtypes(include='str').columns.tolist())
```

### Create Useful Analytical Columns
Age Group

```python
employee_clean["AgeGroup"] = pd.cut(
    employee_clean["Age"],
    bins=[17, 24, 34, 44, 54],
    labels=["18-24", "25-34", "35-44", "45-54"]
)

print(employee_clean["AgeGroup"].value_counts().sort_index())
```

### Salary Band

```python
employee_clean["SalaryBand"] = pd.qcut(
    employee_clean["Salary"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

print(employee_clean["SalaryBand"].value_counts())
```

### Tenure Group

```python
employee_clean["TenureGroup"] = pd.cut(
    employee_clean["YearsAtCompany"],
    bins=[-1, 2, 5, 10],
    labels=["0-2 Years", "3-5 Years", "6-10 Years"]
)

print(employee_clean["TenureGroup"].value_counts().sort_index())
```

### Promotion Gap Group

```python
employee_clean["PromotionGapGroup"] = pd.cut(
    employee_clean["YearsSinceLastPromotion"],
    bins=[-1, 2, 5, 10],
    labels=["0-2 Years", "3-5 Years", "6-10 Years"]
)

print(employee_clean["PromotionGapGroup"].value_counts().sort_index())
```

### Distance Group

```python
employee_clean["DistanceGroup"] = pd.cut(
    employee_clean["DistanceFromHome (KM)"],
    bins=[0, 10, 20, 30, 45],
    labels=["0-10 KM", "11-20 KM", "21-30 KM", "31-45 KM"]
)

print(employee_clean["DistanceGroup"].value_counts().sort_index())
```

### Attrition Flag

```python
employee_clean["AttritionFlag"] = (
    employee_clean["Attrition"] == "Yes"
).astype(int)

print(employee_clean["AttritionFlag"].value_counts())
```
